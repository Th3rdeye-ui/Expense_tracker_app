```python
expenseslist = []  

print("Welcome to Expenses Tracker App")

while True:
    print("\n==== MENU ====")
    print("1. Add Expenses")
    print("2. View Expenses")
    print("3. View Total Expense")
    print("4. Exit")

    choice = input("Please Enter Your Choice: ")

    #1
    if choice == "1":
        print("---------------- Add Your Expense ----------------")
        date = input("Enter the date: ")
        category = input("Enter category (Food, Travel, Makeup, Books): ")
        description = input("Enter description: ")

        try:                             #Error handling
            amount = float(input("Enter the amount: "))
        except ValueError:               #Error handling
            print("Invalid amount. Please enter a number.")
            continue

        expense = {
            "date": date,
            "category": category,
            "description": description,
            "amount": amount,
        }

        expenseslist.append(expense)
        print("\nExpense added successfully")

    # 2.
    elif choice == "2":
        if not expenseslist:
            print("No Expenses Added.")
        else:
            print("\n==== All Expenses ====")
            for count, expense in enumerate(expenseslist, start=1):
                print(
                    f"Expense {count} -> {expense['date']}, "
                    f"{expense['category']}, "
                    f"{expense['description']}, "
                    f"{expense['amount']}"
                )

    # 3. 
    elif choice == "3":
        total = sum(expense["amount"] for expense in expenseslist)
        print("\nTotal Expense =", total)

    # 4. 
    elif choice == "4":
        print("Dhanyavad🙏")
        break

    else:
        print("INVALID CHOICE. TRY AGAIN")
        
```
